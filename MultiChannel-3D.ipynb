```python
import os
from collections import namedtuple

from onekey_algo.classification3d.run_classification3d import main as clf_main3d
from onekey_algo import get_param_in_cwd
# 设置参数
roi_size = [96, 96, 96]
vit_settings = {
        'image_size': roi_size[-1],  # image size
        'frames': roi_size[0],  # number of frames
        'image_patch_size': 16,  # image patch size
        'frame_patch_size': roi_size[0] // 4,  # frame patch size
        'dim': 1024,
        'depth': 6,
        'heads': 8,
        'mlp_dim': 2048,
        'dropout': 0.1,
        'emb_dropout': 0.1}

# 模型存放在哪里
model_root = 'models3D_MultiChannel'
# 你自己组合出来是几个通道
in_channels = 3
# 训练集文件
train_file = r'G:\WHO grading\Zhongyi\模型组件\3D多通道\split_info\train-RND-2.txt'
# 非训练集文件
val_file = r'G:\WHO grading\Zhongyi\模型组件\3D多通道\split_info\val-RND-2.txt'
# 3D 多通道数据存放的位置，nii.gz格式
data_pattern = r'G:\WHO grading\Zhongyi\data\crop\3dnew\3d_multichannel'
# 没有的话会自动创建
labels_file = r'G:\WHO grading\Zhongyi\模型组件\3D多通道\split_info\labels.txt'
# 你喜欢的模型
model_names = ['resnet50', 'resnet18', 'resnet34', 'DenseNet121']
for model_name in model_names:
    params = dict(train=train_file,
                  val=val_file,
                  labels_file=labels_file,
                  data_pattern=data_pattern,
                  j=6,
                  max2use=None,
                  normalize_method='imagenet',
                  model_name=model_name,
                  gpus=[0],
                  roi_size=roi_size,
                  batch_size=4,
                  epochs=96 if model_name in ['ViT', 'SimpleViT'] else 100,
                  init_lr=0.01,
                  optimizer='adam',
                  retrain=None,
                  model_root=model_root,
                  val_interval=1,
                  in_channels=in_channels,
                  iters_verbose=1,
                  model_config={'groups': 1, 'blocks_args': 'efficientnet-b4'},
                  vit_settings=vit_settings,
                  save_per_epoch=True,
                  squeeze_dim=-2,
                  pretrained=False)
    # 训练模型
    Args = namedtuple("Args", params)
    clf_main3d(Args(**params))
```
