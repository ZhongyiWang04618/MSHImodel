```python
import os
from onekey_algo.classification.run_classification import main as clf_main
from collections import namedtuple
from onekey_algo import get_param_in_cwd

# 设置参数
train_f = get_param_in_cwd('train2d_file')
val_f = get_param_in_cwd('val2d_file')
labels_f = get_param_in_cwd('label2d_file')
data_pattern =  get_param_in_cwd('data_pattern2d')

for model_name in get_param_in_cwd('model_names'):
    params = dict(train=train_f,
                  valid=val_f,
                  labels_file=labels_f,
                  data_pattern=data_pattern,
                  j=0,
                  max2use=None,
                  val_max2use=None,
                  batch_balance=False,
                  normalize_method='imagenet',
                  model_name=model_name,
                  vit_settings = {'patch_size': 64, 'dim': 1024, 'depth': 6, 'heads': 16, 'mlp_dim': 768},
                  gpus=[0],
                  batch_size=16,
                  epochs=get_param_in_cwd('epoch', 50),
                  init_lr=0.001,
                  optimizer='adam',
                  retrain=None,
                  model_root=os.path.join(get_param_in_cwd('model_root', 'dl_models'), '2d'),
                  add_date=False,
                  iters_start=0,
                  iters_verbose=1,
                  save_per_epoch=False,
                  pretrained=True)
    # 训练模型
    Args = namedtuple("Args", params)
    clf_main(Args(**params))
```
